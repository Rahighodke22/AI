```python
#Q1 WAP in python to generate random number and print

import random
while True:
    random_int = random.randint(1, 100)
    #print(f"Random Integer between 1 and 100: {random_int}")
    user = int(input("Enter your choice number(between 1 and 100): "))


    if user == random_int:
        print("Execellent!! You guessed the correct number.")
    else:
        print("No match. Better luck next time!")
        print("Correct number was: ",random_int)
    
    play = input("Do you want to play again? (yes/no):")

    if play.lower()!= 'yes':
        print("Thank You for playing")
        break
```

```python
import random

def guess_the_number():
    # Generate a random number between 1 and 100
    secret_number = random.randint(1, 100)
    print("I have chosen a number between 1 and 100. Try to guess it!")

    while True:
        try:
            # Take input from the user
            user_guess = int(input("Enter your guess: "))
            
            # Compare the numbers
            if user_guess > secret_number:
                print("Too high! Try again.")
            elif user_guess < secret_number:
                print("Too low! Try again.")
            else:
                print(f"Congratulations! You guessed the correct number: {secret_number}")
                break # Exit the loop when matched
                
        except ValueError:
            print("Please enter a valid integer.")

# Run the game
guess_the_number()
```
